```python
import gzip
import pickle

from node2vec import Node2Vec
from url_graph import URLGraph
```

```python

graph = URLGraph.read_gexf(
    "graphs/graph_31444_nodes.gexf"
).get_graph()
```

```python

node2vec = Node2Vec(
    graph,
    dimensions=32,
    walk_length=10,
    num_walks=5,
    p=1.0,
    q=1.0,
    workers=4,
)

model = node2vec.fit(
    window=10,
    min_count=1,
    batch_words=256,
)

embeddings = {
    node: model.wv[node].tolist()
    for node in graph.nodes
}
```

```python
with gzip.open("graphs/node2vec_embeddings.pkl.gz", "wb") as file:
    pickle.dump(embeddings, file)
```
