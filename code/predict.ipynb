```python
%load_ext autoreload
%autoreload 2

import gzip
import os
import pickle

import predictor
import url_graph

PATH = os.path.dirname(os.path.abspath(__name__))
```

```python
with gzip.open("graphs/node2vec_embeddings.pkl.gz", "rb") as file:
    embeddings = pickle.load(file)

pred = predictor.Predictor(graph, embeddings=embeddings)
pred.train()
```
