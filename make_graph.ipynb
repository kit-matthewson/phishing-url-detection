```python
df = pd.read_csv('data.csv')
df
```

```python
# Make words column by splitting the url
split_chars = r"./?!&,#$%;="
df['words'] = df['url'].str.split(f"[{split_chars}]")

# Remove empty strings
df['words'] = df['words'].apply(lambda x: [word for word in x if word])  

df
```
